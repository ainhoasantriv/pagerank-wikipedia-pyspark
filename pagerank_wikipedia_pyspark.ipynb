```python
#libraries
from pyspark.sql import DataFrame, functions as F
from pyspark.sql.types import ArrayType, StringType, LongType
from pyspark.sql.functions import col, explode, collect_list, array, lit, size, lower, sum as Fsum, abs as Fabs, array_distinct
import re
```

**1. Load data and check structure**


```python
wikipediaDF = spark.read.parquet(
    "dbfs:/databricks-datasets/wikipedia-datasets/data-001/en_wikipedia/articles-only-parquet")
wikipediaDF.count()
display(wikipediaDF.limit(5))
```

```python
PartialWikipediaDF = wikipediaDF.sample(fraction=0.1, seed=0)
display(PartialWikipediaDF)
PartialWikipediaDF.count()
```

**2. Extract relevant data**

There are several columns, but the relevant information is stored in the follow columns:
  * title: The title of the document
  * id: Id of the document
  * text: The content of the document. The most relevant information here (for the page rank algorithm) are the links to other documents. The  link is enclosed in brackets, and contains the title of the document.

The objective is to obtain a DataFrame with the columns: id, title and links, where the links' column is a list of titles that appear in the article

```python
def parse_links(document_body):
  '''this function works row by row receives a string and returns a list of strings'''
  if document_body is None:
      return []
  document_body = document_body.lower()
  data = re.findall(r'\[\[(.+?)\]\]', document_body)
  return [s.lower() for s in data]
```

Since Spark cannot execute directly Python functions in columns we need UDF.

```python
parse_links_udf = F.udf(parse_links, ArrayType(StringType()))
```

Create the parsed DataFrame

```python
parsedDF = PartialWikipediaDF.select(
    "id",
    F.lower("title").alias("title"),
    parse_links_udf("text").alias("links"))
display(parsedDF.limit(5))
```

**3. Transform Forward Link Table**

The aim is to obtain a ForwardDF with the columns id (the origin page) and links (ids of the paginas de destino)

```python
#this will be the reference table
titleidDF = parsedDF.select(
    F.col("title").alias("link_title"),
    F.col("id").alias("link_id")
)
display(titleidDF.limit(5))
```

```python
#explode the links
explodedDF = parsedDF.select(
    F.col("id").alias("src_id"),
    explode("links").alias("link_title"))
display(explodedDF.limit(5))
```

```python
joinedDF = explodedDF.join(
    titleidDF,
    on="link_title",
    how="inner")
display(joinedDF.limit(5))
```

```python
ForwardDF = (
    joinedDF
    .groupBy("src_id")
    .agg(collect_list("link_id").alias("links"))
    .withColumnRenamed("src_id", "id"))
display(ForwardDF)
```

```python
#add pages without links
ForwardDF = parsedDF.select("id").join(
    ForwardDF,
    on="id",
    how="left"
).withColumn(
    "links",
    col("links").cast("array<long>")  # asegura tipo correcto
)

# replace nulls by empty array
ForwardDF = ForwardDF.withColumn(
    "links",
    F.when(col("links").isNull(), array()).otherwise(col("links"))
)

ForwardDF = ForwardDF.withColumn("links", array_distinct(col("links")))
```

**4. Calculate the number of outputs links**

We want to obtain a DataFrame with the columsn
- id
- count, which is the number of outpots links from the page

```python
count_linksDF = ForwardDF.select(
    col("id"),
    size(col("links")).alias("count"))
'''the function size is better beacuse it avoids sending data to the driver and is faster in bigger datasets'''
count_linksDF = ForwardDF.select(col("id"), size(col("links")).alias("count"))
display(count_linksDF)
```

**5. COnstruct Reverse Table fron Forwar table**

In ReverseDF we want to obtain for each ID page the list of pages that point to this page

```python
ForwardSplitDF = ForwardDF.select(
    col("id").alias("src_id"),
    explode("links").alias("link_id")
)
display(ForwardSplitDF)
```

```python
ReverseDF = (
    ForwardSplitDF
    .groupBy("link_id")
    .agg(collect_list("src_id").alias("links"))
    .withColumnRenamed("link_id", "id")
)
```

```python
#adding the pages that do not have in-links
ReverseDF = parsedDF.select("id").join(
    ReverseDF,
    on="id",
    how="left"
).withColumn(
    "links",
    F.when(col("links").isNull(), array()).otherwise(col("links"))
)

ReverseDF = ReverseDF.withColumn("links", array_distinct("links"))
#display(ReverseDF)
```

```python
# Verifica suma de counts
count_linksDF.agg(Fsum("count")).show()

# Verifica tamaño total de ForwardDF y ReverseDF
ForwardSplitDF.count(), ReverseDF.select(Fsize("links")).agg(Fsum("links")).show()

# Verifica que no haya enlaces duplicados
ForwardSplitDF.groupBy("src_id", "link_id").count().filter("count > 1").show()
```

**6. Initialize PageRank**

In this step the goal is to create a DataFrame with each page and the initial PageRank.


```python
#calculate the total number of pages
N = parsedDF.count()  
print(f"Total pages (N) = {N}")
```

```python
initial_rank = 1.0 / N

PageRankDF = parsedDF.select(
    col("id"),
    lit(initial_rank).alias("pagerank")
)

#display(PageRankDF)
```

**7. Recalculate PageRank table**

The basic PageRank formula (simplified, with damping factor 0.85) is:

$$
PR_i = \frac{1 - d}{N} + d \sum_{j \in L_i} \frac{PR_j}{O_j}
$$

Where:

- \(d = 0.85\) (damping factor)  
- \(N\) = total number of pages  
- \(L_i\) = list of pages pointing to page \(i\) (ReverseDF)  
- \(O_j\) = number of outgoing links from page \(j\) (count_linksDF)  

ReverseDF gives id → links (the pages that point to id)  
count_linksDF gives id → number of outgoing links (out-degree)



```python
#parameters
d = 0.85
N = parsedDF.count()        
max_iter = 20
threshold = 1e-6
```

```python
currentPR = PageRankDF  # initialize

for i in range(max_iter):
    # Explode ReverseDF: one row per link
    explodedRevDF = ReverseDF.select(
        col("id").alias("dest_id"),
        explode("links").alias("src_id")
    )
    # 1. Compute dangling mass (pages with no outgoing links)
    dangling_mass = currentPR.join(
        count_linksDF,
        on="id",
        how="left"
    ).filter(
        col("count") == 0
    ).agg(
        Fsum("pagerank").alias("dangling_mass")
    ).collect()[0]["dangling_mass"] or 0.0

    # Link with actual PageRank
    contribDF = explodedRevDF.join(
        currentPR,
        explodedRevDF.src_id == currentPR.id,
        how="left"
    ).join(
        count_linksDF,
        explodedRevDF.src_id == count_linksDF.id,
        how="left"
    ).select(
        col("dest_id"),
        (col("pagerank") / col("count")).alias("contrib")
    )
    
    # Agrupar por destino para sumar contribuciones
    newPR = contribDF.groupBy("dest_id").agg(Fsum("contrib").alias("sum_contrib"))
    
    # Add damping factor and base
    newPR = newPR.withColumn(
        "pagerank",
        lit((1-d)/N) + lit(d) * col("sum_contrib") + lit(d * dangling_mass / N)
    ).select(
        col("dest_id").alias("id"),
        col("pagerank")
    )
    
    # Pages without in-links: add only the base (dangling nodes)
    missingPR = parsedDF.select("id").join(
        newPR,
        on="id",
        how="left"
    ).withColumn(
        "pagerank",
        F.when(col("pagerank").isNull(), (1-d)/N).otherwise(col("pagerank"))
    ).select("id", "pagerank")
    
    total_pr = missingPR.agg(Fsum("pagerank").alias("sum")).collect()[0]["sum"]
    missingPR = missingPR.withColumn("pagerank", col("pagerank") / total_pr)

    # Compare changes for convergence
    delta = currentPR.join(
        missingPR,
        on="id"
    ).withColumn(
        "diff",
        Fabs(currentPR["pagerank"] - missingPR["pagerank"])  
    ).agg(
        Fsum("diff").alias("total_diff")
    ).collect()[0]["total_diff"]

    
    print(f"Iteration {i+1}, total change = {delta}")
    
    # Update PageRank
    currentPR = missingPR
    
    if delta < threshold:
        print(f"Converged after {i+1} iterations")
        break

# Final result
PageRankDF = currentPR
PageRankDF.show(10, truncate=False)
```

Convert to Pandas and add the titles for the final result.


```python
PageRankPDF = PageRankDF.toPandas()

# add with titles
finalPDF = parsedDF.select("id", "title").toPandas()
finalPDF = finalPDF.merge(PageRankPDF, on="id")
```

**Conclusion Ainhoa Santano**

We implemented the PageRank algorithm using Apache Spark DataFrames to analyze a subset of the Wikipedia dataset. The main objective was to compute the importance of each English Wikipedia Page based on its link structure, while ensuring scalability and efficiency.

The process started by parsing the text of each document to extract the outgoing links using a UDF function. These links were then transformed from page titles into numeric identifiers so that we could work only with IDs and avoiding expensive string operations. With this information we constructed the Forward Link Table, for each document the list of pages it links to. Using this table we built he Reverse Link Table, for each page the list of document that link to it. 

The process countinued by initializing the PageRank values equally for all documents. Then we iteratively recalculated the PageRank using the standard formula with damping factor. In the loop the PageRank values were normalized after each iteration so that the sum remained 1, and the contribution of dandling nodes (pages with np outgoing links) was distributed uniformly across all pages. It stopped when the PageRank vales converged or when the maximum number of iteartions was reached. For the convergence we collected only a single aggregated value which does not affect scalability.

Regarding scalability, all the heavy processing was done using Spark DataFrames (joins, aggregations and transformations). The data was converted to a Pandas DataFrame only at the final step, this allows the solution to work with much larger datasets and to run efficiently on a distributed cluster.

In terms of optimization, we avoided using UDFs when Spark functions were available. For example, we used the size function instead of a UDF to count links. The use of collect() was minimized and applied only to obtain a single value needed to check convergence. This improves performance and reduces memory usage.

In conclusion, the results show different PageRank values which correctly reflect the link structure of the Wikipedia dataset.
